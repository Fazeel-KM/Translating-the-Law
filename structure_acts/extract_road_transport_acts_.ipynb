```python
#  Step 1: Import Required Libraries
import re
import shutil
import os
import json
from zipfile import ZipFile
```

```python
#  Step 2: Define Paths
zip_path = "annotatedCentralActs.zip"
extract_folder = "legal_acts"
filtered_folder = "matching_jsons"
```

```python
#  Step 3: Extract the Zip File
with ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_folder)
```

```python
#  Step 4: Define Target Act Titles
target_acts = {
    "THE INDIAN TOLLS ACT, 1851",
    "THE STAGE-CARRIAGES ACT, 1861",
    "THE INDIAN TOLLS ACT, 1864",
    "THE ROAD TRANSPORT CORPORATIONS ACT, 1950",
    "THE NATIONAL HIGHWAYS ACT, 1956",
    "THE MOTOR VEHICLES ACT, 1988",
    "THE NATIONAL HIGHWAYS AUTHORITY OF INDIA ACT, 1988",
    "THE CONTROL OF NATIONAL HIGHWAYS (LAND AND TRAFFIC) ACT, 2002",
    "THE CARRIAGE BY ROAD ACT, 2007"
}
```

```python
# Step 5: Define a Cleaning Function for Filenames
def clean_filename(text):
    return re.sub(r'[^\w\s-]', '', text).replace(" ", "_")
```

```python
#  Step 6: Filter and Rename Matching JSON Files
os.makedirs(filtered_folder, exist_ok=True)

for root, _, files in os.walk(extract_folder):
    for filename in files:
        if filename.endswith(".json"):
            filepath = os.path.join(root, filename)
            with open(filepath, "r", encoding="utf-8") as f:
                try:
                    data = json.load(f)
                    act_title = data.get("act_title", "").strip().upper()
                    if act_title in target_acts:
                        clean_name = clean_filename(act_title) + ".json"
                        shutil.copy(filepath, os.path.join(filtered_folder, clean_name))
                except Exception as e:
                    print(f"Failed to read {filename}: {e}")
```
